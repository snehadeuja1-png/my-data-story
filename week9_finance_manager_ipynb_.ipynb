```python
data = """Groceries,-55.40
Salary,1500.00
Coffee,-4.50
BAD LINE HERE
Netflix,-15.99
Refund,35.50
Internet,-80
"""

with open("transactions.txt", "w") as file:
    file.write(data)

print("Transaction file created successfully!")
```

```python

def load_transactions(filename):
    try:
        with open(filename, "r") as file:
            lines = file.readlines()
        return [line.strip() for line in lines if line.strip()]
    except FileNotFoundError:
        print("File not found!")
        return []


def process_transactions(lines):
    transactions = []
    failed_lines = 0

    for line in lines:
        try:
            parts = line.split(",")

            if len(parts) != 2:
                raise ValueError("Invalid format")

            description = parts[0].strip()
            amount = float(parts[1].strip())

            transactions.append({
                "description": description,
                "amount": amount
            })

        except ValueError:
            failed_lines += 1
            print("Skipping invalid line:", line)

    print("Invalid lines skipped:", failed_lines)
    return transactions


def show_summary(transactions):
    income = sum(
        t["amount"] for t in transactions
        if t["amount"] > 0
    )

    expenses = sum(
        abs(t["amount"]) for t in transactions
        if t["amount"] < 0
    )

    net_amount = income - expenses

    print("\nPERSONAL FINANCE SUMMARY")
    print("------------------------")
    print("Transactions processed:", len(transactions))
    print(f"Total income: ${income:.2f}")
    print(f"Total expenses: ${expenses:.2f}")
    print(f"Net amount: ${net_amount:.2f}")


def main():
    filename = "transactions.txt"
    lines = load_transactions(filename)
    transactions = process_transactions(lines)
    show_summary(transactions)


main()
```

The hardest part of this project was handling incorrect transaction data without stopping the program. I learned how to use functions to read and process a file and calculate income and expenses. AI helped me understand how try-except statements work and how to skip invalid lines safely. I also learned how the different functions work together to produce a transaction summary.
