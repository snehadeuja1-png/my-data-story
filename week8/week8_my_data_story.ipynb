## Week 8 - Files and Friends

This notebook reads and analyzes my personal text file using Python.
It displays the contents of the file and calculates different statistics
about the information in the file.

```python
# Import the file upload tool from Google Colab
from google.colab import files

# Upload the my_data.txt file
uploaded = files.upload()
```

```python
# Name of the text file we uploaded
filename = "my_data.txt"

# Open the file in read mode
with open(filename, "r") as file:
    content = file.read()
```

```python
# Display the contents of the text file
print("=== MY DATA STORY ===")
print(content)
```

```python
# Split the file contents into separate lines
lines = content.splitlines()

# Count the number of lines
number_of_lines = len(lines)

# Count the number of words
number_of_words = len(content.split())

# Count the total number of characters
number_of_characters = len(content)

# Find the longest line
longest_line = max(lines, key=len)

# Get the first line
first_line = lines[0]

# Get the last line
last_line = lines[-1]

# Display the file statistics
print("=== FILE STATISTICS ===")
print("Number of lines:", number_of_lines)
print("Number of words:", number_of_words)
print("Total characters:", number_of_characters)
print("Longest line:", longest_line)
print("First line:", first_line)
print("Last line:", last_line)
```
